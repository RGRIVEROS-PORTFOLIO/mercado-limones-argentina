## Resumen Ejecutivo

### La pregunta

¿El mercado argentino de autos usados muestra evidencia 
de que la crisis del mercado de 0km — sobrestock, 
descuentos de hasta 25%, guerra de precios — está 
deteriorando la calidad disponible en el mercado secundario?

### La respuesta corta

Sí. Y el momento de mayor deterioro es **ahora**, marzo de 2026.

### Cómo lo sabemos

Se analizaron **1.025.853 transferencias automotrices** 
registradas por la DNRPA entre enero 2023 y marzo 2026, 
en siete puntos temporales seleccionados por su relevancia 
para el contexto macroeconómico argentino.

Se construyó un Índice de Selección Adversa (ISA) que 
combina dos señales de mercado: la antigüedad promedio 
de los vehículos transferidos y la participación de 
vehículos importados bajo el régimen Protocolo 21 
(Mercosur usados).

### El hallazgo central

Marzo 2026 registra el valor más alto del índice en 
toda la serie analizada. Esto coincide exactamente con 
el período de mayor sobrestock reportado por concesionarios 
de 0km y descuentos de hasta el 25% sobre precio de lista.

### Por qué importa

No es una correlación casual. Es un mecanismo identificable: 
cuando el 0km baja de precio, los compradores solventes 
migran hacia el mercado nuevo. En el mercado de usados 
quedan compradores con menor capacidad de verificación, 
que aceptan mayor incertidumbre de calidad a cambio de 
menor precio. El resultado es que la calidad promedio 
disponible en el mercado secundario se deteriora.

## Metodología y Limitaciones

### Qué es este análisis

Un análisis descriptivo basado en datos administrativos 
oficiales (DNRPA), que construye un índice compuesto 
para sintetizar señales de mercado en siete puntos 
temporales seleccionados por criterio teórico.

### Qué NO es este análisis

No es un modelo predictivo. Con n=7 períodos no es 
posible realizar inferencia estadística con significancia 
(regresión, correlación robusta, análisis de lag/lead). 
El volumen de 1.025.853 registros individuales no compensa 
esta limitación: la unidad de análisis del índice es el 
período, no la transacción.

### Limitaciones declaradas

**Muestra no continua:** se analizaron 7 meses específicos, 
no la serie completa 2023-2026. Las tendencias entre 
puntos son inferidas, no medidas directamente.

**Ausencia de datos de precio:** el registro DNRPA no 
contiene información de precio de transacción. El análisis 
se basa en volumen y composición, no en valor monetario. 
Esto impide calcular depreciación directa o magnitud 
económica de la selección adversa.

**Ausencia de historial del vehículo:** no hay información 
sobre accidentes, kilometraje real o estado mecánico. 
La distinción entre hipótesis alternativas sobre Protocolo 21 
(canal de acceso vs. mecanismo de asimetría) no puede 
resolverse únicamente con este dataset.

**Subregistro posible en COM. HAB.:** existe evidencia 
sectorial de que una proporción de operaciones entre 
agencieros se registra bajo categorías de transferencia 
entre particulares. El volumen observado en trámites 
comerciales es un piso, no el total real.

**Ponderación del índice:** el ISA se construyó con dos 
versiones (70/30 y 50/50). Los resultados para los 
períodos extremos (mayor y menor stress) son robustos 
a ambas ponderaciones. Dos períodos intermedios 
(Feb-2024, Oct-2025) son sensibles a la ponderación 
porque sus componentes reflejan mecanismos de mercado 
distintos y no necesariamente convergentes.

### Por qué estas limitaciones no invalidan el análisis

Un análisis exploratorio con limitaciones declaradas 
explícitamente es más confiable que uno que oculta sus 
supuestos. El valor de este trabajo no está en la 
precisión predictiva, sino en identificar un mecanismo 
de mercado consistente con teoría económica establecida, 
usando evidencia empírica verificable y reproducible.

#### Datos posteriores no incorporados: durante el cierre 
de este análisis, DNRPA publicó los archivos correspondientes 
a Abril y Mayo 2026, posteriores a la definición del diseño 
muestral (7 períodos, Ene 2023 - Mar 2026). Estos archivos 
no fueron incorporados para evitar modificar la muestra 
después de haber observado resultados, lo que comprometería 
la validez del análisis de robustez ya ejecutado. Se identifican 
como la primera línea de trabajo futuro (ver sección 
correspondiente).

## Síntesis — Los Tres Hallazgos Centrales

Tres análisis distintos, ejecutados en tres días distintos, 
convergen en la misma conclusión desde ángulos diferentes.

---

### Hallazgo 1 — Taxonomía de marcas (Día 1)

No todas las marcas envejecen igual en el mercado de usados. 
Se identificaron cuatro comportamientos distintos:

| Tipo | Marcas | Mecanismo |
|---|---|---|
| Anti-limón | Toyota, Nissan | Reputación = señal de calidad verificable. Venta rápida, autos jóvenes. |
| Limón clásico | Ford, Fiat, Renault | Modelos discontinuados de alta penetración histórica siguen circulando. |
| Limón premium | Mercedes Benz | Precio de reserva elevado del propietario. Selección adversa de 2do orden. |
| Heterogéneo | Jeep | Dos poblaciones (relanzamiento post-2010 / era IKA-Chrysler) mezcladas. |

**Esto estableció la base:** la antigüedad de un vehículo 
transferido no es aleatoria. Refleja decisiones racionales 
de propietarios que retienen o liberan según el contexto.

---

### Hallazgo 2 — El mecanismo de cascada (Día 2)

Se identificó que la crisis del mercado de 0km no queda 
contenida en ese segmento. Se propaga en tres niveles:

Nivel 1 — Mercado 0km: sobrestock, descuentos hasta 25%

↓

Nivel 2 — Usados Nacional: compradores solventes migran al 0km

↓

Nivel 3 — Protocolo 21: concentra compradores con menor

capacidad de verificación

**Esto fue la evidencia del mecanismo:** no es solo que 
los propietarios retienen autos buenos (Hallazgo 1). 
Es que la demanda misma se reconfigura, empujando a los 
compradores más vulnerables hacia el segmento de mayor 
incertidumbre informacional.

---

### Hallazgo 3 — El Índice de Selección Adversa (Día 3)

La síntesis cuantitativa de los dos hallazgos anteriores. 
El ISA combina antigüedad promedio y participación de 
Protocolo 21 en una sola métrica por período, validada 
con análisis de robustez.

**Resultado:** Marzo 2026 es el punto de mayor stress de 
toda la serie. Diciembre 2024 es el de menor stress, 
explicado por el efecto de liquidación de fin de año 
(Jensen & Meckling).

---

### Por qué los tres hallazgos son una sola historia

La taxonomía (H1) explica POR QUÉ algunas marcas
retienen más que otras.

La cascada (H2) explica CÓMO se propaga la presión
entre segmentos del mercado.

El ISA (H3) MIDE cuándo esa presión es más intensa.

Sin el Notebook 01, no sabríamos que la retención varía por marca.
Sin el Notebook 02, no sabríamos que la presión se transmite 
entre segmentos.
Sin el Notebook 03, no podríamos decir con un número objetivo 
que marzo de 2026 es el peor momento de los últimos tres años.

Los tres noteboks no son tres ejercicios. Son tres piezas 
de un mismo argumento.

## Recomendaciones — Bajadas a Tierra

El análisis tiene valor solo si se traduce en decisiones 
concretas para los actores reales del mercado. A continuación, 
recomendaciones diferenciadas según el rol.

---

### Para el agenciero (vendedor de usados)

**El problema que muestran los datos:**
Cuando el 0km baja de precio, tu cliente potencial con 
mejor perfil crediticio se va al concesionario. Te queda 
el comprador que no califica para ese crédito o no tiene 
el diferencial de precio. Esa demanda acepta menos, pero 
también exige menos verificación — lo cual no es una 
ventaja, es una señal de que está comprando con apuro 
o con menos opciones.

**Lo que el ISA muestra para marzo de 2026:**
Es el peor momento de los últimos tres años en términos 
de calidad disponible para vender. Tu inventario bueno 
está compitiendo de manera desfavorable contra el 0km 
con descuento. Es matemáticamente más difícil moverlo 
al precio que vale.

**Recomendación concreta:**
No reduzcas precio de tus mejores unidades para igualar 
al 0km — esa pelea no la podés ganar. Diferenciá por 
verificación: ofrecé informe de dominio, servicio reciente 
documentado, e idealmente una inspección de un tercero. 
Es la única ventaja competitiva real que tenés frente 
a un 0km con descuento: la transparencia que el comprador 
no encuentra fácil en otro lado.

---

### Para el comprador de usados

**Lo que indican los datos para este momento del mercado:**
Los autos buenos están retenidos por sus dueños. Lo que 
hay disponible en volumen son las unidades que alguien 
necesita liquidar — no necesariamente las mejores.

**Sobre Protocolo 21 específicamente:**
Su participación está en el máximo histórico de la serie 
(42%). Esto no significa que sean autos malos. Significa 
que no podés evaluarlos con la misma información que un 
nacional con historial DNRPA accesible.

**Recomendación concreta:**
Si podés esperar la normalización del mercado de 0km, 
esperá — los vendedores de autos buenos volverán a 
ofertar cuando dejen de competir contra descuentos 
artificiales. Si necesitás comprar ahora y el presupuesto 
te lleva a un Protocolo 21, destiná entre el 3% y el 5% 
del valor del vehículo a una verificación profesional 
antes de cerrar la operación. Es la forma más económica 
de reducir la asimetría de información que el mercado 
no te da gratis.

---

### Para una financiera de créditos automotrices (el caso de negocio original)

**Lo que el análisis sugiere para política de riesgo:**

El origen del vehículo (`automotor_origen`) es una variable 
de riesgo relevante y subestimada. Un crédito sobre un 
Protocolo 21 enfrenta mayor incertidumbre sobre el estado 
real del vehículo que uno sobre un nacional con historial 
DNRPA verificable.

**El ISA como indicador de timing de mercado:**

En períodos de ISA alto (como marzo 2026), el inventario 
de usados disponible tiene mayor probabilidad de ser de 
menor calidad relativa. Esto sugiere ajustar criterios 
de tasación o exigir inspección adicional en períodos 
de stress identificados por el índice.

**Recomendación concreta:**

Incorporar el origen del vehículo y un proxy del ISA 
mensual como variables en el scoring de riesgo crediticio 
para préstamos sobre vehículos usados, no solo el modelo 
y año del vehículo como se hace habitualmente.

## Líneas de Trabajo Futuro

Este proyecto fue diseñado con un scope acotado y un 
presupuesto de tiempo definido. Las siguientes líneas 
quedan identificadas como continuación natural, en orden 
de prioridad.

---

### 1. Incorporar Abril y Mayo 2026

DNRPA publicó estos archivos durante el cierre de este 
análisis. Permitirían verificar si la tendencia de marzo 
2026 (mayor stress de la serie) se sostiene, se profundiza, 
o revierte. Es la actualización más simple de ejecutar 
y la de mayor valor inmediato.

### 2. Cruce con datos de precios — Infovalores 

El análisis actual mide volumen y composición, no valor 
monetario. Cruzar los registros de transferencia con las 
tablas de valuación de ACARA permitiría calcular 
depreciación real por marca y origen, y cuantificar 
económicamente la magnitud de la selección adversa — 
no solo su dirección.

### 3. Serie completa vs. muestra puntual

El diseño actual usa 7 meses seleccionados por criterio 
teórico. Procesar la serie mensual completa 2023-2026 
(28 archivos) permitiría aplicar análisis de series de 
tiempo con n suficiente para inferencia estadística real: 
descomposición de tendencia y estacionalidad, y detección 
de quiebres estructurales.

### 4. Validación de la Hipótesis A vs. B sobre Protocolo 21

El dataset DNRPA no permite distinguir si Protocolo 21 
funciona como canal de acceso a vehículos de calidad o 
como mecanismo de asimetría de información. Esto requeriría 
una fuente adicional: encuestas a compradores, datos de 
siniestralidad por origen, o reportes de servicios de 
verificación vehicular.

### 5. Análisis a nivel de transacción individual

El ISA opera a nivel de período (n=7). Con 1.025.853 
registros individuales sería posible construir un modelo 
de clasificación o regresión a nivel de transacción 
(por ejemplo, predecir antigüedad esperada según marca, 
origen y provincia), aprovechando el volumen real de 
datos disponible para inferencia estadística robusta.

### 6. Comparación interprovincial

El análisis actual no explotó la columna 
`registro_seccional_provincia` más allá del diagnóstico 
inicial. Comparar el ISA por provincia permitiría 
identificar si la selección adversa es uniforme en 
el país o se concentra en mercados específicos.

## Conclusión

### La tesis central

El mercado argentino de vehículos usados (2023-2026) exhibe 
un mecanismo de selección adversa consistente con Akerlof (1970), 
que no es constante sino que se intensifica de forma medible 
en respuesta a shocks específicos del mercado de 0km. 
El momento de mayor intensidad detectado en este análisis 
es marzo de 2026 — el presente, no el pasado.

### Lo que este proyecto demuestra

Que es posible tomar un registro administrativo público, 
sin curar y sin metadata analítica, y construir desde cero:

- Una limpieza documentada y reproducible
- Una taxonomía original de comportamiento de mercado por marca
- La identificación de un mecanismo de propagación entre 
  tres segmentos del mercado automotriz
- Un índice compuesto validado con análisis de robustez
- Recomendaciones aplicables a tres actores reales del mercado

### Lo que este proyecto no pretende ser

No es un modelo predictivo. No cuantifica en pesos la magnitud 
de la selección adversa. No resuelve si Protocolo 21 es 
oportunidad o riesgo — identifica que es ambas cosas según 
el comprador y declara la pregunta abierta donde corresponde.

### La frase que resume el proyecto

> Cuando el mercado de autos nuevos baja de precio, 
> el mercado de usados no se beneficia. Se vacía de calidad.

Eso no es una opinión. Es lo que muestran 1.025.853 
transferencias registradas por la DNRPA entre enero de 2023 
y marzo de 2026.
